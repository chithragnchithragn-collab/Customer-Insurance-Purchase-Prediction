```python
plt.figure(figsize=(6,4))
sns.countplot(x='Purchased', data=df)
plt.title('Distribution of Purchased')
plt.xlabel('Purchased(0 = No, 1 = Yes)')
plt.ylabel('Number of Customers')
plt.show()
```

```python
plt.figure(figsize=(7, 4))
sns.histplot(data=df, x='Age', kde=True, bins=15)
plt.title('Customer Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(7, 4))
sns.histplot(data=df, x='EstimatedSalary', kde=True, bins=20)
plt.title('Customer Estimated Salary Distribution')
plt.xlabel('Estimated Salary')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Age', y='EstimatedSalary', hue='Purchased',palette='Set1')
plt.title('Age vs. Estimated Salary-- Purchase Analysis')
plt.xlabel('Age')
plt.ylabel('Estimated Salary')
plt.show()
```
